Изучаем BAC_HIEN_CONSTRUCTION_SAFETY_2024, версия 8, в формате COCO JSON. Проверим изображения и рамки, распределения классов и связанные сцены. По результатам сформируем рабочий датасет и правила разметки.


## Окружение и пути

Параметры источника и пути читаем из общего конфига.


```python
from pathlib import Path

import yaml

ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "config.yaml").is_file()
)
config = yaml.safe_load((ROOT / "config.yaml").read_text(encoding="utf-8"))
settings = config["datasets"]["safety"]
SEED = config["seed"]
RAW_DIR = ROOT / settings["raw_dir"]
REPORT_DIR = ROOT / settings["report_dir"]

print("Данные:", RAW_DIR)
print("Результаты:", REPORT_DIR)
print("Seed:", SEED)
```
